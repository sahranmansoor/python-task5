```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('sales_data.csv')

print("DataFrame Shape (rows, columns):", df.shape)
display(df.head())

print("\nMissing values per column:")
display(df.isnull().sum())

df['Sales'] = df['Sales'].fillna(0)
```

```python
high_sales = df[df['Sales'] > 300]
print(f"Rows with Sales > 300: {len(high_sales)}")

print("\nUsing iloc[] to get the first 3 rows and first 2 columns:")
display(df.iloc[:3, :2])

print("\nUsing loc[] to get specific columns by name:")
display(df.loc[:3, ['Category', 'Sales']])
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales = category_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
category_sales.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Total Sales by Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
